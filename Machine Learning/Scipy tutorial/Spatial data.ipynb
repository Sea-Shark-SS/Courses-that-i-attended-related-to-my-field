```python
# =====================================================================
# STEP 1: IMPORT NECESSARY LIBRARIES
# =====================================================================
import numpy as np
import matplotlib.pyplot as plt
# Importing key spatial classes and metrics from the SciPy Spatial module [00:05:45]
from scipy.spatial import Delaunay, ConvexHull, KDTree
from scipy.spatial.distance import euclidean, cosine

print("="*60)
print("             SCIPY SPATIAL DATA DEMONSTRATION                ")
print("="*60)

# =====================================================================
# STEP 2: DELAUNAY TRIANGULATION [00:04:40]
# =====================================================================
print("\n--- SECTION 1: DELAUNAY TRIANGULATION ---")

# Define spatial coordinate points matching the video layout [00:06:34]
points = np.array([
    [2, 4],
    [3, 4],
    [3, 0],
    [2, 2],
    [4, 1]
])

# Compute Delaunay triangulation [00:07:44]
tri = Delaunay(points)

# Render the triangulation plot [00:09:06]
plt.figure(figsize=(6, 4))
plt.triplot(points[:, 0], points[:, 1], tri.simplices.copy(), color='blue', label='Triangulation')
plt.scatter(points[:, 0], points[:, 1], color='red', zorder=5, label='Points')
plt.title("Delaunay Triangulation Example")
plt.legend()
plt.grid(True)
plt.show()

print("Delaunay Triangulation Simplices:\n", tri.simplices)


# =====================================================================
# STEP 3: CONVEX HULL ENVELOPE [00:16:14]
# =====================================================================
print("\n" + "="*60)
print("--- SECTION 2: CONVEX HULL ---")

# Define an expanded coordinates matrix for envelope calculation [00:17:38]
hull_points = np.array([
    [2, 4], [3, 4], [3, 0], [2, 2], [4, 1],
    [5, 0], [3, 1], [1, 2], [0, 2], [0, 4]
])

# Compute Convex Hull boundary layers [00:19:40]
hull = ConvexHull(hull_points)

# Draw the boundary hull path layout [00:21:42]
plt.figure(figsize=(6, 4))
plt.scatter(hull_points[:, 0], hull_points[:, 1], color='blue', label='All Points')
for simplex in hull.simplices:
    plt.plot(hull_points[simplex, 0], hull_points[simplex, 1], 'k-', linewidth=2)
plt.title("Convex Hull Boundary Envelope")
plt.legend()
plt.grid(True)
plt.show()

print("Vertices forming the Convex Hull perimeter:\n", hull.vertices)


# =====================================================================
# STEP 4: MULTI-DIMENSIONAL KD-TREES QUERIES [00:26:46]
# =====================================================================
print("\n" + "="*60)
print("--- SECTION 3: KD-TREE SPATIAL QUERIES ---")

# Set up spatial lookup reference arrays [00:29:56]
tree_data = np.array([
    [1, -1],
    [2, 3],
    [-2, 3],
    [2, -3]
])

# Build the search optimization tree layout [00:30:58]
tree = KDTree(tree_data)

# Query the tree structure for the nearest neighbor to point [1, 1] [00:31:35]
target_query_point = [1, 1]
distance, index = tree.query(target_query_point)

print(f"Target Query Point: {target_query_point}")
print(f"Nearest Neighbor Distance: {distance:.4f}")
print(f"Nearest Neighbor Data Array Index: {index} (Coordinates: {tree_data[index]})")


# =====================================================================
# STEP 5: MATH SPATIAL DISTANCE METRICS [00:35:42]
# =====================================================================
print("\n" + "="*60)
print("--- SECTION 4: EUCLIDEAN VS COSINE DISTANCES ---")

# Create two coordinate vectors to compare metrics [00:40:29]
v1 = np.array([1, 0])
v2 = np.array([10, 2])

# Measure standard Euclidean step displacement distance [00:41:04]
e_dist = euclidean(v1, v2)
# Measure Angular Cosine divergence distance tracking [00:44:29]
c_dist = cosine(v1, v2)

print(f"Vector 1: {v1} | Vector 2: {v2}")
print(f"1. Straight-Line Euclidean Distance Measure: {e_dist:.4f}")
print(f"2. Angular Orientation Cosine Distance Measure: {c_dist:.4f}")

print("\n" + "="*60)
print("All Spatial Operations Successfully Executed in Colab Space.")
print("="*60)
```
