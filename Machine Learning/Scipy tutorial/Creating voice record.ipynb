```python
# =====================================================================
# STEP 1: DOWNLOADING DEPENDENCIES VIA PIP TERMINAL
# =====================================================================
# !pip install sounddevice scipy

# =====================================================================
# STEP 2: IMPORT CORE AUDIO EXTRACTION PACKAGES
# =====================================================================
import sounddevice as sd            # Hardware peripheral audio interaction [00:01:22]
from scipy.io.wavfile import write  # Submodule to export arrays to .wav file [00:03:08]

print("="*65)
print("             PYTHON VOICE RECORDER SYSTEM INITIALIZED             ")
print("="*65)

# =====================================================================
# STEP 3: DEFINE THE CORE SOUND RECORDER ENGINE FUNCTION [00:04:42]
# =====================================================================
def voice_recorder(duration_seconds, output_filename):
    """
    Captures live audio arrays via system microphone inputs and writes them to a file.

    Parameters:
    duration_seconds (int): Total seconds to record [00:05:01]
    output_filename (str): Output string file destination path [00:05:08]
    """
    # Configuration profiles
    sampling_frequency = 44100  # High fidelity audio sampling standard (44.1kHz) [00:06:43]
    audio_channels = 2          # Configuration index for 2-Channel Stereo Output [00:08:38]

    print(f"\n[STATUS] >>> Recording Started... Speak into your microphone now.")
    print(f"[INFO]   >>> Duration Set: {duration_seconds} seconds.")

    # Trigger the hardware extraction process buffer [00:06:14]
    recording_buffer = sd.rec(
        int(duration_seconds * sampling_frequency),
        samplerate=sampling_frequency,
        channels=audio_channels
    )

    # Introduce blocking mechanism until recording buffer matches duration length [00:09:23]
    sd.wait()

    print("[STATUS] >>> Recording Completed successfully.")

    # Write structural float/integer arrays into an uncompressed .wav format [00:09:47]
    write(output_filename, sampling_frequency, recording_buffer)
    print(f"[SUCCESS] >>> Sound data compiled and exported as: '{output_filename}'\n")

# =====================================================================
# STEP 4: TRIGGER PRODUCTION TEST SUITE RUNNER [00:11:12]
# =====================================================================
if __name__ == "__main__":
    # Specify the name of the file to save the recording [00:11:34]
    target_file = "my_voice_sample.wav"

    # Set recording time window length to 10 seconds [00:11:27]
    record_duration = 10

    # Execute the recording script [00:11:20]
    voice_recorder(record_duration, target_file)

    print("="*65)
    print("      Process Completed. Check your directory for output files.")
    print("="*65)
```
